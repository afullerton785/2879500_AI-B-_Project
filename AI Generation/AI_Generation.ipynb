### How Generative AI Works

Google Gemini and a plethora of other image generating AI models operate in the same way. These models are trained on massive datasets of images and learn the statistical patterns and distributions within that data. Rather than copying existing images, the model learns relationships between visual features such as shapes, textures, and colours, and uses them to produce an output.

During the training of these models, algorithms such as transformers, diffusion models, and variational autoencoders are used to teach a neural network to capture these patterns as a compact, structured representation in latent space (Sadat et al., 2024). Once training is complete, the model can take new data (the prompts we give it) and generate a new piece of data using the features from its training.

The basic workflow involves:

Training Data -> Algorithm -> Model -> New Input Data -> Generated Output (Cetinic and She, 2022)

In this scenario for example the model takes the original illustration, then to modernise it uses images and text it has been trained on to create an output that should resemble the prompt; in this case the modernised versions of the illustrations from the pamphlet. 

Although the AI model generates the images, the process of this generation is very much a collaboration between the human and the AI model. This is because without the input of the human the AI has nothing to make, and the human must be able to select the correct prompts and refine them to create a result that is exactly what was needed (Dang et al., 2022). For example in this project, the human selects the illustrations to be modernised, designs the prompts that will be used to generate images and refines them where necessary, then finally uses these images to influence their own recreation of the illustration.

I elected to include this in the final project rather than fixing it as I believe it to be a perfect example of where AI image generation still struggles and where a human touch could be deemed necessary to help with the process.
